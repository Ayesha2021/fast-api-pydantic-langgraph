# Optional Fields

```python
from typing import Optional
```

```python
# Testing optional fields
class NewEmployee(BaseModel):
    name:str
    age:int
    city:str
    DOB:datetime
    salary:Optional[float]=None # default value None
    isActive: Optional[bool]=False # default value is False
    
```

```python
newEmp =NewEmployee(name="John Doe",age=30,city="Nebraska",DOB=datetime.now())
print(newEmp)
```

```python
# Automatic TypeCasting is done example below for salary float type casting
newEmp1 =NewEmployee(name="John Doe",age=30,city="Nebraska",DOB=datetime.now(),salary=9876)
print(newEmp1)
```

```python
# Model with list of values --how validation done
class Library(BaseModel):
    libId:str
    bookTitle:List[str] # List of String corresponding to book names in Library
    bookCount: int
    
```

```python
#create Library
#library = Library('book1',['abc','xyz'],2)
library =Library(
    libId='lib1',
    bookTitle =['Cat in the mat','wimpy kid'],
    bookCount =2)
print(library)
```

```python
# Type casting supported by pydantic tuple type casted to list
library1 =Library(
    libId='lib1',
    bookTitle =('Cat in the mat','wimpy kid'), # providing tuple instead of List
    bookCount =2)
print(library1)
```

```python
# if you give dict instead of list data validation error from pydantic
# Type casting supported by pydantic tuple type casted to list
library2 =Library(
    libId='lib1',
    bookTitle = {'name':'Cat in the mat',value:'wimpy kid'}, # providing set/dictionary instead of List
    bookCount =2)
print(library2)
```

```python
try:
    invalidDataLibrary =Library(
    libId='lib1',
    bookTitle =2, # providing integer instead of List
    bookCount =2)
    print(invalidDataLibrary)
except ValueError as e:
    print(e)
    
```

# Adding Constraints on top of Type hint using Field,you can add default values,aliases,validation rules.

```python
class LineItem(BaseModel):
    name:str =Field(min_length=2,max_length=50) # by default type is str you can add additional contraint lik min len,max len
    price:float=Field(gt=0,le=4000)
    quantity:int =Field(gt=0)
#Validate 
lineItem =LineItem(
    name='iphone',
    price=567.0,
    quantity=1
)
print(lineItem)
```

```python
class LineItem(BaseModel):
    name:str =Field(min_length=2,max_length=50) # by default type is str you can add additional contraint lik min len,max len
    price:float=Field(gt=0,le=4000)
    quantity:int =Field(gt=0)
#Validate 
invalidLineItem =LineItem(
    name='i',
    price=567.0,
    quantity=0
)
print(invalidLineItem)
```

## Fields with default Values


```python
class Person(BaseModel):
    name:str=Field(...,description='Full Person Name ...in first positional argument denote optional') # ... in first positinal argument means name is required
    age:int= Field(default=14,description='Person Age default to 14')
    email:str = Field(default_factory=lambda: 'person@testing.com',description='default email address')
```

```python
#printing as is all correct values
person = Person(name='testperson')
    
    #do not provide age defaults to 14 and do not provide email let is take default
print(person)
```

```python
#printing as is all correct values
personValid = Person(name='testperson',age=34,email='testing@pydatic.com')
    
    #provide all desired values.
print(personValid)
```

```python
#you can print schema of any model
#print(person.schema()) deprecated
print(person.model_json_schema())
```
